## Step 6: Error handling and cost awareness

```python
import time
```

```python
def get_completion_with_retry(prompt, max_retries=3):

    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    for attempt in range(max_retries):
        try:
            response = client.messages.create(
                model="claude-haiku-4-5-20251001",
                max_tokens=5,
                messages=messages
            )

            return response.content[0].text

        except anthropic.APITimeoutError as e:
            print(f"Timeout occurred (attempt {attempt + 1}/{max_retries})")

            if attempt == max_retries - 1:
                raise Exception(
                    "API call failed after multiple retries."
                ) from e

            # Wait before retrying
            time.sleep(2)
```

```python
get_completion_with_retry("just testing function", max_retries=3)
```
